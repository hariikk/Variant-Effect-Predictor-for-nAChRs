## 9. Alternative Architectures to Try (Beyond "Just Add Features")

### 9.1 MissION-Style Dual-Channel Architecture

```
┌─────────────────────┐     ┌──────────────────────┐
│  ESM-2 650M         │     │  52 Engineered        │
│  Differenced Embed  │     │  Features              │
│  (1280 → 50 PCA)    │     │  (physicochem + struct │
│                     │     │   + subst + positional)│
└─────────┬───────────┘     └───────────┬────────────┘
          │                             │
          ▼                             ▼
    ┌──────────┐                ┌──────────────┐
    │  MLP     │                │  MLP         │
    │  (64→32) │                │  (64→32)     │
    └────┬─────┘                └──────┬───────┘
         │                             │
         └──────────┬──────────────────┘
                    ▼
            ┌──────────────┐
            │  Concatenate │
            │  + Classifier│
            │  (LightGBM / │
            │   XGBoost /  │
            │   MLP head)  │
            └──────────────┘
                    │
                    ▼
              GOF / LOF
```

**Why:** The dual-channel pattern is proven by MissION (ROC-AUC 0.925 on ion channels). It lets pLM embeddings and structured features learn separate representations before merging.

**Comparison to current approach:** Currently we just concatenate all 52 features and feed to a single model. The dual-channel approach adds learned nonlinear combinations within each feature group before merging.

### 9.2 ClearVariant-Style Joint Sequence Modeling

```
┌──────────────────────┐
│  WT Sequence         │
│  "M D S G L ..."     │
└──────────┬───────────┘
           │
           ▼
    ┌─────────────────┐
    │  ESM-2 Encoder  │
    │  (shared weights)│
    └────────┬────────┘
             │
    ┌────────┴────────┐
    │                 │
    ▼                 ▼
┌─────────┐    ┌──────────────┐
│ WT emb  │    │  Var emb     │
│ (at pos)│    │  (at pos)    │
└────┬────┘    └──────┬───────┘
     │                │
     └───────┬────────┘
             ▼
    ┌──────────────────┐
    │  Self-Attention  │
    │  over [WT; Var;  │
    │   CLS] tokens    │
    └────────┬─────────┘
             ▼
    ┌──────────────────┐
    │  CLS → Classifier│
    └────────┬─────────┘
             ▼
         GOF / LOF
```

**Why:** ClearVariant achieves GOF precision 0.814 by explicitly comparing WT and variant sequences through attention. The model learns which residue pairs drive the GOF/LOF distinction.

**Implementation note:** This requires encoding full sequences (or windows), not just per-position embeddings. Higher compute cost but potentially much better GOF detection.

### 9.3 PreMode-Style Two-Stage Transfer Learning

**Stage 1 — Pretraining (done once, can reuse public weights):**
- Train on ClinVar + HGMD pathogenic/benign variants (~148K)
- Learn general "how damaging is this mutation" (parameter **r**)
- Use SE(3)-equivariant GNN on AlphaFold structures

**Stage 2 — Fine-tuning on nAChR:**
- Freeze pretrained weights for parameter **r**
- Fine-tune only the direction head (parameter **θ**) on our 351 GOF/LOF variants
- ~30 min on single GPU

**Why:** This is how PreMode achieves AUC ~0.75 with few dozen variants per gene. It leverages the fact that GOF/LOF direction is a *refinement* on top of general pathogenicity.

---

### 9.4 FSFP-Style MAML + Ranking

**Idea:** Instead of binary classification (GOF vs LOF), frame as **ranking:** "rank these variants from most GOF to most LOF." Ranking loss provides more signal per variant than binary cross-entropy.

**Implementation sketch:**
1. Pretrain with MAML on diverse ProteinGym DMS tasks
2. Fine-tune with LoRA on nAChR variants using pairwise ranking loss
3. At test time: rank variants by predicted score → threshold for binary classification

**Why:** FSFP showed ranking loss consistently beats regression for few-shot protein tasks. Our 351 variants might benefit from the richer training signal of ranking.

---

### 9.5 Subunit Phylogeny Multi-Task Learning

**Idea:** Instead of one model for all subunits (with subunit one-hot), train separate task for each subunit but with a **sequence-similarity kernel** that lets data-rich subunits (CHRNA1, CHRNE) lend strength to data-poor ones (CHRNA6: 18 rows, CHRNA2: 9, CHRNA5: 5).

**Implementation:** Already sketched in features.ipynb §3.10. The new finding is that this is **exactly the architecture used by the K⁺-channel MTL paper** (EBioMedicine 2022) — validated on an ion channel family with the same uneven-per-gene-data problem.

## 7. New Methods: Small-Data Strategies (Critical for 351 Variants)

### 7.1 Fitness Translocation — Augment from Orthologs (Mialland et al., bioRxiv 2024)

**This is the single most actionable small-data method for our project.**

**Method:**
1. Encode variant and WT sequences from a **homologous protein** (e.g., mouse nAChR) using ESM-2
2. Compute the embedding-space "mutation offset" (variant_embedding − WT_embedding)
3. Apply that offset to the **target protein's** WT embedding → synthetic augmented variant
4. Merge synthetic variants with real experimental data for training

**Key results:**
- Consistent improvement across IGPS, GFP, SARS-CoV-2 spike proteins
- **Especially effective at very small training sizes (as low as 45 variants)**
- Works with remote homologs sharing only **35% sequence identity**
- Includes selection algorithms to identify which homolog datasets actually help — prevents negative transfer

**How we'd use it:**
1. Use mouse nAChR data (already in the pipeline via `load_multi_species_dataset()`) as the homolog source
2. Encode mouse WT and variant sequences with ESM-2
3. Compute mutation offsets in embedding space
4. Apply offsets to human WT sequences → synthetic human variants
5. Merge with real human data → train classifier
6. Compare: human-only vs. augmented on LSO-CV

**Code:** `github.com/adrienmialland/ProtFitTrans`

---

### 7.2 FSFP — Few-Shot Protein Fitness Prediction (Zhou et al., Nature Comms 2024)

**What:** Combines **meta-transfer learning (MAML) + learning to rank (LTR) + LoRA**. With only **20 labeled single-site mutants**, boosted Spearman correlation by ~0.1 across 87 DMS datasets. Validated in wet-lab (Phi29 DNA polymerase, 25% increase in positive rate).

**Key insights:**
- **Ranking-based loss consistently beats regression** for few-shot protein tasks
- MAML + LoRA compatible with ESM-1v, ESM-2, and SaProt
- GEMME used to construct auxiliary training tasks

**How we'd use it:**
1. Use MAML to pretrain on general variant effect prediction tasks (from ProteinGym)
2. Fine-tune with LoRA on our 351 nAChR variants using ranking loss
3. Compare ranking vs. classification performance

**Code:** Based on the paper description. **Publication:** `nature.com/articles/s41467-024-49798-6`

---

### 7.3 MERGE — DCA + Stacked Regression (Illig et al., J. Chem. Inf. 2024)

**What:** Combines Direct Coupling Analysis (DCA) from MSAs with stacked regression. Low-dimensional sequence-specific encoding minimizes overfitting vs. high-dimensional pLM embeddings. **Works effectively with 50–500 labeled sequences.**

**Why relevant:** Our 351 variants are in the sweet spot. MERGE's low-dimensional approach may outperform pLM embeddings when data is this small.

---

### 7.4 LoRA Fine-Tuning of ESM-2 (Schmirler et al., Nature Comms 2024)

**What:** Systematic benchmarking of ESM-2 fine-tuning strategies. Key finding: **LoRA achieves comparable improvement to full fine-tuning with 4.5× acceleration.** Explicitly recommended for small datasets.

**Implementation:**

## 6. New Methods: Structural & Biophysical Features for Ion Channels

### 6.1 Pore-Axis Distance — Strongest Single Structural Predictor

**Source:** Brunger et al. (2023, *Brain*). Analysis of 3,049 pathogenic and 12,546 population variants across **30 ion channel structures** with **163 structural features**.

**Key finding:** Pore residue properties and **proximity to the pore axis** were the single most significantly enriched features for pathogenic variants. The pore-axis distance gradient correlated with:
- Seizure age of onset
- Cognitive performance  
- **Differential GOF vs. LOF effects**

**Implementation for nAChR:**
1. Define pore axis from pentamer symmetry (5-fold axis through M2 helices)
2. Compute perpendicular distance for every variant residue to the pore axis
3. Bin into: <5Å (pore-lining), 5–10Å (near-pore), >10Å (peripheral)
4. Add binary "pore-lining" feature
5. Cross-reference with M2 9' position hydrophobic gate residues

**Structures to use:** 6CNJ (α4β2), 6PV7 (α3β4), 7EKT/7EKI (α7), 7KOX (muscle), 9DMG

**Estimated effort:** Low-Medium. Requires pentamer assembly structures (available for cryo-EM structures, not AlphaFold monomers).

---

### 6.2 N-AChR Allosteric Pathway Position (Novel Feature)

The nAChR signal transduction pathway is well-characterized:
```
Agonist binding (orthosteric site) → Cys-loop → β10-TM1 linker → TM2-M3 loop → M2 helix (gate)
```

**Proposed feature:** Ordinal pathway position per variant:
- 0 = Orthosteric / ligand-binding site (Loops A–F)
- 1 = Cys-loop (conserved disulfide)
- 2 = β10–TM1 linker (ECD-TMD coupling)
- 3 = TM2–M3 loop (gating hinge)
- 4 = M2 pore-lining helix (gate itself)
- 5 = Other TMD / ICD / ECD (non-pathway)

**Why this should help:** Mutations at different pathway positions have mechanistically *opposite* effects. The CMS paper (Engel, Ohno & Sine, 2002) already documented this: fast-channel (LOF) = binding-site mutations; slow-channel (GOF) = M2 pore mutations. An allosteric pathway position encodes this domain knowledge as a feature.

---

### 6.3 RSA-Stratified Feature Interactions

**Why:** A buried charge change is far more damaging than the same change at the surface. Current features treat all positions equally.

**Implementation:**
- Bin RSA into: buried (<20%), partially exposed (20–50%), exposed (>50%)
- Create interaction terms: RSA_bin × BLOSUM62, RSA_bin × Grantham, RSA_bin × charge_change
- Add binary: "buried charge change" (RSA<20% AND charge-altering substitution)
- Add binary: "Cys-loop disulfide breaker" (mutation at Cys128 or Cys142 in α subunits)
- Add binary: "H-bond breaker" (WT residue participates in ≥2 H-bonds per DSSP)

---

### 6.4 FoldX ΔΔG on nAChR Structures

**Why:** ΔΔG (stability change) has **not been directly used** as a feature in any published GOF/LOF classifier — this is a genuine research gap.

**Hypothesis:** LOF = destabilizing (positive ΔΔG), GOF = neutral or stabilizing for the activated state. For nAChR specifically: compute ΔΔG from BOTH open-state and closed-state structures if available — the *difference* may predict direction.

**Implementation:**
```bash
# FoldX BuildModel on all 351 variants using cryo-EM pentamer structures
FoldX --command=BuildModel --pdb=7QKO.pdb --mutant-file=variants.txt
```
- Use multimeric (pentameric) structures, not monomers
- Cross-validate FoldX ΔΔG with known GOF/LOF labels
- Consider "Foldetta" (mean of FoldX + Rosetta) for better calibration

**Tools:** FoldX 5.1 (free academic), Rosetta flex_ddg, GeoDDG (best-in-class DL), StaB-ddG (>1000× faster than FoldX)

---

### 6.5 CAVER Pore Geometry Features

**What:** CAVER 3.0 computes pore radius profiles, bottleneck radius, and bottleneck residue identities from static structures. ChannelsDB 2.0 links pore annotations with UniProt mutation data.

**Features to extract:**
1. Pore bottleneck radius (Å)
2. Distance of variant residue to bottleneck
3. Binary: "bottleneck residue" (residue at narrowest pore constriction)
4. Binary: "in M2 helix" (pore-lining)
5. Distance to 9' gate residue in 3D space
6. Local pore hydrophobicity at variant position

**Tool:** `github.com/loschmidt/caver` (PyMOL plugin) or `MDAnalysis.analysis.hole2`

---

### 6.6 B-Factor & Conformational State Features

**B-factor normalized by structure:** High B-factor regions (flexible loops) → mutations more tolerated. Low B-factor core/pore residues → highly constrained.

**Conformational state difference features:** For subunits with both open and closed structures:
- Compute each structural feature from BOTH states
- Include the DIFFERENCE (open − closed) as features
- Hypothesis: mutation destabilizing open > closed = LOF; destabilizing closed > open = GOF

**Additional structural feature concepts:**
- **HSE-up/HSE-down ratio** — captures directional burial asymmetry (pore-facing vs. lipid-facing)
- **Membrane burial depth** (Z-coordinate relative to membrane midplane)
- **Electrostatic ring proximity** — nAChR has charged rings at pore entrances
- **Interface proximity** — minimum distance to any other subunit (binding site, gating interface)
- **Multi-radius contact counts** at 8Å and 12Å (extends existing cbeta_density)

## 5. New Methods: Protein Language Model Approaches

### 5.1 ESM-2 Embeddings as Features (Highest Priority)

**What:** Per-residue embeddings from ESM-2 capture sequence context, evolutionary information, and biophysical properties learned from 65M+ UniRef sequences — without needing MSAs.

**Key finding from 2024–2025 literature:**
- **ESM-2 650M is the optimal size for small datasets** (<10,000 observations) — matches or beats the 3B/15B variants on transfer tasks (*Scientific Reports*, 2025, PMC11601519)
- **Mean pooling outperforms** max pooling, BOS tokens, and PCA for downstream classification
- **Fine-tuning (even last 2 layers) consistently beats frozen embeddings** for supervised tasks
- **LoRA fine-tuning achieves comparable improvement with 4.5× speedup** (Schmirler et al., *Nature Communications*, Aug 2024)
- Zero-shot masked marginal scoring alone has **Spearman ρ ≈ 0.44–0.45** on ProteinGym — useful as a feature, not standalone

**Implementation plan:**
```python
# HuggingFace model ID
MODEL = "facebook/esm2_t33_650M_UR50D"
# For each variant:
# 1. Mask the mutated position in wildtype sequence
# 2. Forward pass → get logits for all AAs at masked position
# 3. LLR = log P(mutant_aa) - log P(wildtype_aa)  → single scalar feature
# 4. Per-residue embedding at mutation site → 1280-dim vector
# 5. Difference embedding = embed(mutant_seq) - embed(wt_seq) at mutation site
```

**Feature set to extract:**
1. ESM-2 LLR score (1 feature — zero-shot fitness)
2. ESM-2 embedding at mutation position, WT sequence (1280 features)
3. ESM-2 embedding at mutation position, VAR sequence (1280 features)
4. Difference embedding (1280 features)
5. PCA-reduced difference embedding (e.g., 50 features — to avoid 1280-dim explosion with 351 samples)

**Estimated effort:** Medium (2–5 days). HuggingFace download ~2.5GB. CPU-feasible for single-sequence zero-shot scoring. Minutes for all 351 mutations.

### 5.2 VespaG — GEMME→ESM-2 Knowledge Distillation (Schlensok et al., 2024)

**What:** Shallow feed-forward NN (single hidden layer, 660K params) mapping ESM-2 (3B) embeddings → variant effect scores. Trained via knowledge distillation using **GEMME as teacher** on ~39M human proteome variants. No MSAs needed at inference.

**Key numbers:**
- Mean Spearman 0.48 on ProteinGym (217 DMS assays) — matches SaProt, TranceptEVE, AlphaMissense
- **Entire proteome's mutational landscape in <30 min on a laptop**
- GEMME-teacher distillation strategy is a template: use MSA-based evolutionary scores as a starting feature, build supervised classifier on top

**Code:** `github.com/jschlensok/vespag`

---

### 5.3 SaProt — Structure-Aware pLM (ICLR 2024 Spotlight)

**What:** Extends ESM-2 with a "structure-aware vocabulary" — each residue = concatenated token: amino acid ID (uppercase) + Foldseek-encoded 3D structural state (lowercase, 20 discrete structure alphabet letters). Trained on ~40M AlphaFold2-predicted structures.

**Why it matters for nAChR:**
- nAChRs undergo dramatic conformational changes (open/closed)
- SaProt embeddings encode **both sequence identity AND local backbone conformation**
- This naturally captures the structural context that distinguishes pore-lining from surface residues
- Complement to our existing DSSP-based structural features

**Code:** `github.com/westlake-repl/SaProt`

---

### 5.4 ProteinMPNN-ddG — Stability from Inverse Folding (2024, NeurIPS MLSB)

**What:** Repurposes inverse folding models (ProteinMPNN, ESM-IF) for stability prediction **without retraining**. Key insight: ΔΔG ≈ log-odds(folded context) − log-odds(single residue context), where single-residue context approximates the unfolded state.

**Performance:** 77% precision on stabilizing mutations (up from 66%). ~10,000 residues/second on V100 GPU. Full human proteome in ~30 minutes.

**Why this matters for GOF/LOF:** LOF variants tend to destabilize protein core → positive ΔΔG. GOF variants may be neutral or stabilizing. ΔΔG could help separate the two classes, and this method is computationally cheap.

**Code:** `github.com/PeptoneLtd/proteinmpnn_ddg`

### 5.5 ThermoMPNN / ThermoMPNN+ (Kuhlman Lab, 2024 PNAS)

**What:** GNN trained on ProteinMPNN backbone to predict ΔΔG. ThermoMPNN+ integrates ESM embeddings via "latent transfusion."

**Performance:** 68% precision on stabilizing mutations. Spearman 0.552, R² 0.196.

**Code:** `github.com/Kuhlman-Lab/ThermoMPNN`

## 4. New Methods: GOF/LOF Classification (Direct Comparators)

### 4.1 ClearVariant — Current GOF SOTA (Ha et al., 2025)

**Citation:** Ha, Kim, Kwon, Chung, Han et al. "Learning sequence to predict gain- or loss-of-function variants." Research Square preprint, 2025. DOI: rs-6705195.

**What:** Attention-based deep learning on ESM-2 embeddings with **joint reference-variant sequence modeling** — explicitly compares wild-type and variant sequences using a self-attention interpretability layer.

**Why it matters for us:**
- **GOF precision: 0.814** (vs. LoGoFunc's 0.578 — massive jump)
- **GOF F1: 0.778** (vs. LoGoFunc's 0.670)
- Co-evolved residue pairs drive predictions — directly relevant for nAChR inter-subunit coupling
- The joint reference-variant architecture is a design pattern we should adopt

**How we could use it:**
1. Adopt the joint WT+var sequence modeling pattern: instead of just features at the mutation site, explicitly encode both sequences and let attention find relevant residue pairs
2. Feed ESM-2 embeddings for both WT and variant sequences, plus our 52 engineered features
3. Add a self-attention interpretability layer to understand which residue interactions drive predictions (useful for thesis)

**Code:** Preprint only, no public code yet. But the architecture is described and reproducible.

---

### 4.2 PreMode — SE(3)-GNN with Two-Stage Transfer (Zhong et al., 2025)

**Citation:** Zhong, Zhao, Zhuang, Chung & Shen. "PreMode predicts mode-of-action of missense variants by deep graph representation learning." *Nature Communications*, August 2025. DOI: 10.1038/s41467-025-62318-4.

**What:** SE(3)-equivariant graph attention transformers. Two-parameter model: **r** (distance from wildtype = pathogenicity magnitude) and **θ** (direction = GOF vs LOF). Two-stage transfer: (1) genome-wide pathogenicity pretraining on ~148K variants, (2) gene-family-specific GOF/LOF fine-tuning.

**Key findings relevant to nAChR:**
- GOF variants: enriched on **protein surfaces**, **low-pLDDT regions** (50-90), **alpha-helices** in ion channels
- LOF variants: greater impact on **folding energy (ΔΔG)**, cluster in **structured core regions**
- Transfer learning effective with as few as **dozens of labeled variants per gene**
- In ion channels: GOF enriched in alpha-helices (unlike general proteome where GOF enriched in disordered regions)

**How we could use it:**
1. **Two-stage transfer:** Pretrain a model on general ClinVar pathogenic/benign → fine-tune on nAChR GOF/LOF
2. This is the single most transferable idea — our 351 variants fit the "few dozen" scenario
3. Use PreMode's pretrained weights if available, or implement the architecture from scratch
4. Caveat: Full pretraining needs ~250GB RAM + 4× A40 GPUs. Transfer learning on smaller hardware ~30 min on single GPU.

**Code:** `github.com/ShenLab/PreMode` | Data: `huggingface.co/gzhong/PreMode.Data`

### 4.3 MissION — Closest Existing Tool (Gies et al., 2025)

**Citation:** Gies, Alishbayli, Tiesinga & Martens. "Functional Effect Predictions For Ion Channel Missense Variants Using a Protein Language Model." medRxiv, October 2025. Published *Journal of Human Genetics*, 2026.

**What:** Dual-channel architecture: ESM-2 (3B) embeddings cropped to ±200 residues around variant + MLP processing 68 annotated features (sequence, structure, residue-based, HPO terms, GO terms). Trained on 3,176 experimentally characterized GOF/LOF variants.

**Architecture insight:** The dual-channel design (pLM embeddings + structured features) is the **proven template** for ion channel GOF/LOF. We should adopt this pattern directly:
- Channel 1: ESM-2 650M embeddings at mutation site (differenced: mutant − wildtype)
- Channel 2: Our 52 engineered features (physicochemical + structural + substitution + positional)
- Merge → classifier (LightGBM, XGBoost, or shallow NN)

**Code:** `github.com/SynapticaNL/MissION` | Zenodo: `zenodo.org/records/20729069`
