### Slide 2 — What is VEP

**On screen:** VEP = Variant Effect Predictor; the flow *Variant → VEP model → Effect (LOF/GOF)*; What / Why columns.

**Say:**
"VEP stands for Variant Effect Predictor. The idea is simple: you give it a protein
variant — usually a missense change, where one amino acid is swapped for another — and it
predicts what that change does to the protein's function. Why do we need this? Sequencing
has revealed millions of variants in people, but only a tiny fraction have ever been tested
in the lab, and those experiments — like electrophysiology — are slow and expensive. So a
computational predictor lets us triage variants: flag the ones most likely to matter, and
help guide diagnosis and treatment."

**Transition:** "This project builds directly on an earlier VEP for a related ion channel, ENaC."   

### Slide 3 — Related Work: VEP for ENaC

**On screen:** ENaC = epithelial Na⁺ channel; the people in the title (Alexander Hagg, Noah Plingen, Karen Lizet Luján López, Mike Althaus); pipeline; "Our basis".

**Say:**
"Our work builds on a previous project — a variant effect predictor for ENaC, the
epithelial sodium channel. That project originated with Mike Althaus, with the experimental
expertise of Mike and Karen Lizet Luján López, and the predictor itself was developed
together with Alexander Hagg and Noah Plingen. Their pipeline is the template we follow:
take a curated set of variants, turn them into engineered features, and train classical
machine-learning models. We reuse that same interpretable-feature philosophy and then add
a few features specific to nAChRs. So this is our foundation — we're adapting a proven
approach to a new channel family."

**Transition:** "So what is that channel family? Let's look at the nicotinic acetylcholine receptors."   

### Slide 4 — Nicotinic Acetylcholine Receptors

**On screen:** Structure (pentameric ligand-gated channel, 16 subunits), Role (synaptic transmission — CNS, neuromuscular), Disease; pentamer diagram.

**Say:**
"Nicotinic acetylcholine receptors — nAChRs — are ligand-gated ion channels. 'Ligand-gated'
means the channel opens when a signalling molecule — acetylcholine — binds to it; that
binding is the trigger. Structurally they're pentamers: five subunits arranged around a
central pore, and humans have 16 different subunits that combine in different ways. Their
job is fast synaptic transmission, both in the central nervous system and at the
neuromuscular junction, where nerve meets muscle. When variants disrupt them you get
diseases like congenital myasthenic syndromes and epilepsy, and they're linked to nicotine
dependence. The key point for us: a single amino-acid change can either weaken the channel
or make it overactive."

**Transition:** "And that distinction — weaken versus overactivate — is exactly the problem we're solving."   

### Slide 5 — The Problem: Direction of Effect

**On screen:** Key question (LOF vs GOF); two columns; "Direction ≠ pathogenicity".

**Say:**
"Here's the core question: given a missense variant, does it cause a *loss of function* or a
*gain of function*? Loss of function means reduced or absent activity — the channel doesn't
conduct, or doesn't reach the cell surface. Gain of function is the opposite — increased or
prolonged activity, an over-excitable channel. This matters because direction is *not* the
same as pathogenicity. Most predictors just say 'this variant is damaging.' But the
direction tells you the *mechanism* — and crucially, the right treatment is often opposite
depending on whether it's a loss or a gain. Give the wrong-direction drug, and you can make
the patient worse."

**Transition:** "So why can't we just use the existing variant predictors for this?"   

### Slide 6 — Existing Predictors

**On screen:** Generic VEPs score pathogenicity only — no direction; comparison table.

**Say:**
"The popular tools — AlphaMissense, ESM, PolyPhen, REVEL, CADD — all score pathogenicity on
a single axis: how damaging is this variant. None of them output a direction, so they can't
separate loss from gain. They're still useful, though — we can feed their scores in as input
features. The table lists the methods that actually *do* predict direction, which are our
real comparators: funNCion for sodium and calcium channels, LoGoFunc genome-wide, PreMode,
and MissION for ion channels. The bottom row is us — specialised to the nAChR family, using
interpretable engineered features."

**Transition:** "So let me state our objective clearly."   

### Slide 7 — Objective

**On screen:** Goal + three sub-goals.

**Say:**
"Our goal is a nAChR-specific predictor of the *direction* of effect — loss versus gain —
built from interpretable, biologically meaningful features. To get there, three sub-goals:
first, curate a clean, labelled database of nAChR variants; second, engineer four families
of features; and third, benchmark several machine-learning models using a careful,
imbalance-aware cross-validation."

**Transition:** "Let's start with the data."   

### Slide 8 — Dataset

**On screen:** 351 variants, 16 subunits, LOF/GOF binary, filtering, imbalance bar chart.

**Say:**
"Our dataset is curated from the literature: 351 missense substitutions across the
16-subunit nAChR family, each labelled as loss or gain of function. We keep only single
amino-acid substitutions, and drop ambiguous cases and variants with no net effect. One
important property: the classes are imbalanced — 218 losses versus 133 gains, roughly 62 to
38. That imbalance is exactly why we later evaluate with macro-F1 and stratified
cross-validation rather than plain accuracy — accuracy would reward a model that just always
guesses 'loss'."

**Transition:** "Next: how do we turn a variant into something a model can learn from — the features."   

### Slide 9 — Feature Engineering

**On screen:** 52-dimensional vector, four feature groups table.

**Say:**
"Each variant is converted into a 52-dimensional vector, grouped into four feature families.
Physicochemical features capture the chemistry change. Substitution features capture how
unusual the amino-acid swap is. Positional features capture where in the protein, and which
subunit. And structural features capture the residue's 3-D environment. Everything here is
interpretable and biologically motivated — there are no black-box embeddings in this
baseline, which means we can always explain *why* a prediction was made. Let me walk through
each group."

**Transition:** "First, the physicochemical features."   

### Slide 10 — Physicochemical Features (24)

**On screen:** "What" column + the 8 property scales.

**Say:**
"These ask: does the swap change the residue's chemistry? We use eight property scales —
things like hydrophobicity, polarity, size, and charge — and for each scale we store three
numbers: the value for the original amino acid, the value for the new one, and the
difference between them. Eight scales times three is 24 features. The difference, delta, is
variant minus wild-type, so it captures both how big the change is and in which direction.
The intuition: a big chemical jump — say, a small hydrophobic residue becoming a large
charged one — is more likely to disrupt the channel."

**Transition:** "Second, the substitution features."   

### Slide 11 — Substitution Features (3)

**On screen:** BLOSUM62 (raw + normalised), Grantham distance.

**Say:**
"These ask: how rare or unusual is this particular swap? We use two well-established
measures. BLOSUM62 is an evolutionary score — how often this substitution shows up across
related proteins; a high score means a common, conservative swap, a negative score means a
rare, likely disruptive one. We use both the raw and a normalised version. Grantham distance
is a chemical distance between amino acids — zero means identical, and the maximum, 215, is
the most extreme swap, like cysteine to tryptophan. So a low BLOSUM, or a high Grantham,
points to a more drastic change."

**Transition:** "Third, the positional features."   

### Slide 12 — Positional Features (17)

**On screen:** normalised position, subunit one-hot; "Why it matters".

**Say:**
"These capture *where* the mutation sits and in *which* subunit. There's one number for the
position along the protein, scaled to a zero-to-one range, and then 16 'one-hot' values — a
simple on/off flag marking which of the 16 subunits this variant belongs to. That's 17
features. Why does it matter? Functionally critical regions — like the pore lining and the
acetylcholine-binding loops — cluster at characteristic positions, and the tendency toward
loss or gain can differ between subunits. So the model can learn subunit-specific patterns."

**Transition:** "And fourth, the structural features."   

### Slide 13 — Structural Features (8)

**On screen:** per-residue descriptors (RSA, B-factor, DSSP, Cβ density + HSE), "No structure → AlphaFold"; PDB table.

**Say:**
"These describe the residue's local 3-D environment in the folded channel. Per residue we
compute: solvent accessibility — is it buried or exposed; the B-factor, a measure of local
flexibility; the secondary structure — helix, sheet, or loop; and packing measures like
C-beta density and half-sphere exposure. The table on the right lists the experimental
structures we used — cryo-EM and X-ray structures for the muscle receptor and the main
neuronal receptors. And the line 'no structure → AlphaFold' means this: six of the subunits
have no experimentally solved structure, so for those we use an AlphaFold model instead —
AlphaFold is DeepMind's deep-learning system that predicts a protein's 3-D shape directly
from its sequence. That lets us still extract these structural features even where no
laboratory structure exists."

**Transition:** "Now that we have the features, here's the overall pipeline."   

### Slide 14 — Pipeline Overview

**On screen:** Data → Features → Models → Nested CV → Prediction (LOF/GOF).

**Say:**
"This ties it all together. We start from the curated variant database, turn each variant
into its 52 features, feed that into the models, train and tune them with nested
cross-validation, and out comes the predicted direction — loss or gain. It's a clean,
classical machine-learning pipeline, deliberately interpretable from end to end."

**Transition:** "Let me say a bit more about the models and how we evaluate them."   

### Slide 15 — Methodology

**On screen:** four models; nested CV bullets; nested-CV diagram.

**Say:**
"We benchmark four classical models — logistic regression, a support vector machine, random
forest, and LightGBM, a gradient-boosting method. For evaluation we use nested
cross-validation, which is the rigorous way to do this. The outer loop holds out a fold of
data purely for testing; the inner loop, on the remaining data, tunes the model's
hyper-parameters using Optuna. The crucial point is that the test fold is never used during
tuning — so we don't fool ourselves with optimistic scores. We repeat the whole thing over
five random seeds for stability, and our headline metric is macro-F1, because it treats the
minority gain-of-function class fairly despite the imbalance."

**Transition:** "One challenge with this kind of project is simply data — so we're working to expand it."

# Methodology — deep dive & Q&A prep

*(Background so you can field questions. You will NOT say all of this in the talk — it's your
safety net. The analogies are the important part; read this a few times.)*

**30-second mental model.** Machine learning here = teaching a computer to guess **LOF vs GOF**
from **52 numbers** per variant. We show it 351 examples where we already know the answer
(that's **training**); it finds patterns linking the numbers to the answer. Then we test it on
variants it has **never seen**, to check it actually *learned something general* instead of just
memorising.

### 1. The setup: supervised binary classification

- Each variant is a **row of 52 numbers** (the *features*, `X`) plus a **label** (`y` = LOF or GOF).
- **Supervised** = for the training examples we know the correct answer.
- **Binary classification** = exactly two possible answers (LOF or GOF).
- **Training (fitting)** = the model adjusts its internal numbers so its guesses match the known
  answers.
- **Prediction** = feed a new variant's 52 numbers in, get a predicted label + a confidence
  (probability).
- The whole game is **generalisation** (work on new variants), not **memorisation** (that failure
  is called *overfitting* — more below).

### 2. The four models, in plain English

We compare four, going from **simple** to **complex**. If a simple one does as well as a complex
one, the complexity isn't buying us anything — and using several shows the result isn't a fluke of
one algorithm.

**Logistic Regression (LR)** — the simple baseline.
- *What:* computes a **weighted sum** of the 52 features, then squashes it to a probability
  between 0 and 1 (the *sigmoid* function); above 0.5 -> GOF, below -> LOF. It draws a **straight
  line** (flat boundary) between the two classes.
- *Analogy:* a scoring sheet — each feature gets a weight (importance + direction), you add them up.
- *Main knob:* `C` (regularisation — how hard to penalise large weights so it doesn't overfit).
- *Pros:* fast, and **interpretable** (you can read which features push toward GOF). *Cons:* only
  straight-line boundaries.

**Support Vector Machine (SVM) with RBF kernel.**
- *What:* finds the boundary that leaves the **widest gap (margin)** between the two classes. The
  **RBF kernel** lets that boundary be **curvy** instead of straight.
- *Analogy:* pushing a flexible wall between two crowds, as far from both as possible.
- *Main knobs:* `C` (margin vs mistakes trade-off) and `gamma` (how wiggly the boundary is).
- *Pros:* good with non-linear patterns. *Cons:* less interpretable; needs feature scaling.

**Random Forest (RF).**
- *What:* a **forest of decision trees**. Each tree asks a chain of yes/no questions about the
  features (e.g. "is Grantham distance > 100?") and votes LOF or GOF; the forest **averages the
  votes**.
- *Analogy:* a committee of simple flowcharts. Each tree sees a random slice of the data and
  features, so their individual mistakes cancel out.
- *Main knobs:* number of trees, max depth, min samples per leaf, features per split.
- *Pros:* captures non-linear effects and feature interactions, robust, no scaling needed.
  *Cons:* larger, less interpretable than LR.

**LightGBM (gradient-boosted trees)** — usually the strongest on tabular data like ours.
- *What:* also trees, but built **one after another**, where each new tree focuses on the
  **mistakes** the previous trees made. This is *boosting*.
- *Analogy:* a relay team where each new member specialises in fixing the previous members' errors.
- *Main knobs:* learning rate, number of leaves, max depth, number of trees, regularisation.
- *Pros:* typically the most accurate here. *Cons:* more knobs to tune; can overfit if not tuned.

### 3. Parameters vs hyperparameters (common question)

- **Parameters** = learned *automatically* during training (LR's weights, a tree's split points).
  You never set these by hand.
- **Hyperparameters** = settings you choose *before* training that control *how* it learns (tree
  depth, learning rate, `C`, `gamma`). The model can't learn these itself — you must **search** for
  good values.
- *Analogy (baking):* parameters = how the batter reacts in the oven (happens on its own);
  hyperparameters = the oven **temperature and time** you set — wrong settings ruin the cake.

### 4. Tuning the hyperparameters with Optuna

- *Problem:* we don't know the best hyperparameters in advance.
- *Solution:* **try many combinations, measure each, keep the best.** We run **50 trials** per fold.
- **Optuna** is a library that searches **smartly** — it learns from earlier trials which settings
  look promising (its "TPE sampler") instead of blindly trying every combination (grid search) or
  purely random ones.
- **Key rule:** tuning is judged on *validation* data (the inner loop, below) — **never** on the
  final test data.

### 5. Why cross-validation (not a single train/test split)?

- One split into train/test is **lucky or unlucky** — your score depends on which variants happened
  to land in the test set (and 351 is small).
- **k-fold CV:** split the data into **5 equal parts** ("folds"). Train on 4, test on the 1 left
  out; **rotate** so each fold is the test set once; **average** the 5 scores. Much more reliable.
- **Stratified** = each fold keeps the same LOF:GOF ratio (~62:38), so no fold is accidentally
  all-LOF.
- We also repeat the whole thing with **5 different random shuffles ("seeds")** and average, so the
  result doesn't hang on one particular shuffle.

### 6. Nested cross-validation — the heart of it (expect questions here)

**The trap it avoids:** if you use the *same* data to (a) pick the best hyperparameters **and**
(b) report your score, the model has indirectly "seen" the test data -> your score comes out
**better than reality**. That's called **data leakage / optimistic bias**.

**Nested CV separates those two jobs with two loops:**
- **Outer loop (5 folds):** each fold is held out **once as a pure test set** — used *only* to
  score, never for tuning.
- **Inner loop (5 folds):** *inside* the outer training data, run another 5-fold CV + Optuna to pick
  the best hyperparameters.
- Then **refit** the model on the full outer-training set with those hyperparameters, and score
  **once** on the untouched outer-test fold.
- Repeat over **5 seeds** -> 25 honest test scores -> report **mean +/- standard deviation**.

**Exam analogy:** the inner loop is your **practice exams** (you use them to decide how to study);
the outer loop is the **real final exam you never saw while studying**. If you'd tuned on the final
exam, your grade would be meaningless.

**One-line answer to memorise:** *"Nested CV keeps hyperparameter tuning and performance estimation
on separate data, so our reported numbers aren't inflated by tuning on the test set."*

### 7. How we measure success (metrics)

**Confusion matrix** — 4 boxes (taking GOF as the "positive" class):
- **TP** = GOF correctly called GOF; **FP** = LOF wrongly called GOF;
- **FN** = GOF wrongly called LOF; **TN** = LOF correctly called LOF.

- **Accuracy** = fraction correct. *Misleading here:* since 62% are LOF, a lazy model that **always
  says LOF** scores 62% while being useless. So we don't rely on accuracy.
- **Precision (GOF)** = of the ones we *called* GOF, how many really were. ("When I say GOF, am I
  right?")
- **Recall (GOF)** = of the *real* GOFs, how many we caught. ("Do I find the GOFs?")
- **F1** = the **harmonic mean** of precision and recall — only high when **both** are high.
- **Macro-F1** = the F1 of the LOF class and the F1 of the GOF class, **averaged with equal
  weight**. Because both classes count equally, the model **can't win by ignoring the minority GOF
  class** — that's why we use it under imbalance.

> *Honesty note for you:* the current code actually reports the **F1 of the GOF class specifically**
> (the hard minority class). Same spirit as macro-F1. If someone presses on the exact metric, say:
> *"F1 focused on the minority gain-of-function class."*

### 8. Overfitting, imbalance, scaling (loose ends)

- **Overfitting** = the model **memorises** the training examples (including noise) and then fails on
  new data. Tell-tale sign: great on training, poor on test. Our guards: cross-validation, simple
  models, regularisation, and never tuning on the test fold.
- **Underfitting** = model too simple to capture the pattern (poor on *both* train and test).
- **Class imbalance** (218 LOF / 133 GOF) -> handled by **stratified folds** + **macro-F1** so the
  minority isn't ignored.
- **Feature scaling:** LR and SVM care about the **numeric range** of features (they use distances /
  weighted sums), so we standardise features (a `RobustScaler`); tree models (RF, LightGBM) **don't
  need it**. Crucially the scaler is **fit on the training fold only**, then applied to the test fold
  — fitting on all data would leak information.

### 9. Anticipated questions — with short answers you can say out loud

**Q: Why classical ML instead of deep learning / neural networks?**
A: "With only 351 examples, deep networks overfit badly. Classical models on engineered features are
more reliable on small tabular data, and they're interpretable."

**Q: What's your baseline — how do you know the model learned anything?**
A: "Always predicting the majority class gives ~62% accuracy. Our models have to beat that, and
macro-F1 has to beat a trivial classifier."

**Q: Why nested CV instead of a normal train/test split?**
A: "The dataset is small and we also tune hyperparameters. Nested CV gives an unbiased estimate by
never tuning on the fold we test on."

**Q: What's the difference between validation and test data?**
A: "Validation (inner loop) picks the hyperparameters; test (outer loop) is only for the final score
and is never seen during tuning."

**Q: Which hyperparameters did you tune?**
A: "Model-specific, via Optuna — e.g. `C` and `gamma` for the SVM; learning rate, number of leaves,
depth and regularisation for LightGBM; depth and number of trees for random forest."

**Q: Why macro-F1 and not accuracy?**
A: "The classes are imbalanced; accuracy rewards always predicting the majority. Macro-F1 weights
both classes equally, so the minority GOF class actually counts."

**Q: Isn't 351 variants too few?**
A: "Yes, it's a real limitation — that's exactly why we use simple models, rigorous cross-validation,
and are augmenting the data with mouse variants."

**Q: How do you avoid data leakage?**
A: "The test fold is untouched during tuning; even the feature scaler and the hyperparameter search
only ever see training data."

**Q: Which model performed best?**
A: "They're close; the boosted trees (LightGBM) and logistic regression are the strongest. LR is the
most interpretable, LightGBM usually the most accurate." *(adjust to your actual result.)*

**Q: Can you interpret the model / which features matter?**
A: "Yes — that's the payoff of engineered features. Logistic-regression weights and tree
feature-importances show which features drive predictions, e.g. substitution scores and position."

**Q: What is a random seed?**
A: "A number that fixes the random shuffling so runs are reproducible; we average over 5 seeds so the
result doesn't depend on one lucky split."

**Q: Why not ROC-AUC?**
A: "We can report it too — AUC measures ranking quality across thresholds. We emphasise
macro-F1 / minority-class F1 because we care about correctly calling *both* directions under
imbalance."

**Q: Explain the SVM kernel in one sentence.**
A: "The RBF kernel measures how similar two variants are and lets the decision boundary curve, so
classes that aren't separable by a straight line can still be split.\"
